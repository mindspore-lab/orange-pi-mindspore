# safetensors 转 pytorch bin（float16）

在 Windows 本地运行，把 safetensors 转成 pytorch bin 格式，同时转换为 float16。

**前置依赖：**
```
pip install safetensors torch
```

````python
!pip install safetensors torch
````

````python
from safetensors.torch import load_file
import torch
import os
import json

model_dir = './wongwian-micro-instruct'
safetensors_path = os.path.join(model_dir, 'model.safetensors')
bin_path = os.path.join(model_dir, 'pytorch_model.bin')
config_path = os.path.join(model_dir, 'config.json')

print(f'读取: {safetensors_path}')
state_dict = load_file(safetensors_path)

print('转换权重为 float16...')
state_dict_fp16 = {k: v.to(torch.float16) for k, v in state_dict.items()}

print(f'写入: {bin_path}')
torch.save(state_dict_fp16, bin_path)

with open(config_path, 'r') as f:
    config = json.load(f)
config['torch_dtype'] = 'float16'
with open(config_path, 'w') as f:
    json.dump(config, f, indent=2)
print('✓ config.json 已更新为 float16')

print(f'文件大小: {os.path.getsize(bin_path) / 1024 / 1024:.1f} MB')
````
